 Código para las iteraciones del problema 5b


```python
import math

# Función
def f(x):
    return math.exp(x) - x**2 + 3*x - 2

# Intervalo dado
a = 0
b = 1

# Tolerancia dentro de la cual se desea que éste la solución
tol = 1e-5

# test de cambio de signo (Teorema del Valor Intermedio)
if f(a) * f(b) > 0:
    print("No hay cambio de signo en el intervalo.")
    
# iteraciones... 
else:
    iteracion = 0
    
    while (b - a) / 2 > tol:
        iteracion += 1
        c = (a + b) / 2
        
        print(f"Iteración {iteracion}:")
        print(f"a = {a:.6f}, b = {b:.6f}, c = {c:.6f}, f(c) = {f(c):.6f}")
        
        # test de cambio de signo
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
    
    raiz = (a + b) / 2
    print("\nRaíz aproximada =", round(raiz, 6))
    print("Número de iteraciones =", iteracion)

```
