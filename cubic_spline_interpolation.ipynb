Spline cúbico natural del ejercicio 17 sección 3.5.

17. Construya un spline cúbico natural para aproximar $f(x) = \cos(\pi x)$ al usar los valores dados por $f(x)$ en $x = 0,\; 0.25,\; 0.5,\; 0.75,\; 1.0$.  Integre el spline sobre $[0,1]$ y compare el resultado con $\int_0^1 \cos(\pi x)\,dx = 0$. 
Use las derivadas del spline para aproximar $f'(0.5)$ y $f''(0.5)$. Compare estas aproximaciones con los valores reales.

Para el ejercicio 19 de la sección 4.1 solo se requiere el valor aproximado de la segunda derivada valuada en $0.5$

```python
# librerias a usar 
import numpy as np  # Cálculo numérico
import sympy as sp  # cálculo simbólico

import pandas as pd # Tablas

# Para mostrar expresiones en formato LaTeX 
from IPython.display import display, Math
```

```python
# Datos 

## Nodos
x = np.array([0.0, 0.25, 0.5, 0.75, 1.0])

# Valores f(x) = cos(pi x)
a = np.cos(np.pi * x)

n = len(x) - 1  # Número de intervalos

# Mostrar tabla con los datos
tabla = pd.DataFrame({
    "x": x,
    "f(x)": a
})

tabla
tabla
```

```python
# Cálculo de h_i (distancias entre nodos consecutivos)

# Arreglo de ceros para almacenar los valores de h
h = np.zeros(n)

# se recorre cada intervalo entre nodos consecutivos  para ir hallando las distancias correspondientes
for i in range(n):
    # h[i] representa la distancia entre x[i+1] y x[i]
    h[i] = x[i+1] - x[i]

# Se muestran los valor de cada h_i en un vector
print("h =", h)
```

```python
# Cálculo de alpha_i  

# Arreglo de ceros para almacenar los valores de alpha
alpha = np.zeros(n)

# Se recorre desde i = 1 hasta n-1  
for i in range(1, n):
    # Fórmula de alpha_i del sistema del spline cúbico
    alpha[i] = (3/h[i])*(a[i+1] - a[i]) - (3/h[i-1])*(a[i] - a[i-1])

# Se muestran los valores
print("alpha =", alpha)
```

```python
# Eliminación hacia adelante del sistema tridiagonal

# Se crean vectores auxiliares del algoritmo de spline cúbico
l = np.zeros(n+1)   # diagonal principal modificada
mu = np.zeros(n+1)  # multiplicadores
z = np.zeros(n+1)   # término auxiliar

# Condiciones iniciales
l[0] = 1
mu[0] = 0
z[0] = 0

# Recorrido hacia adelante 
for i in range(1, n):
    # Construcción del sistema tridiagonal reducido
    l[i] = 2*(x[i+1] - x[i-1]) - h[i-1]*mu[i-1]

    # Factor de eliminación
    mu[i] = h[i] / l[i]

    # Término independiente modificado
    z[i] = (alpha[i] - h[i-1]*z[i-1]) / l[i]

# Condiciones finales
l[n] = 1
z[n] = 0

# Resultados
print("l =", l)
print("mu =", mu)
print("z =", z)
```

```python
# Sustitución hacia atrás

c = np.zeros(n+1)
b = np.zeros(n)
d = np.zeros(n)

# Condición final
c[n] = 0

# Sustitución hacia atrás
for j in range(n-1, -1, -1):
    c[j] = z[j] - mu[j]*c[j+1]
    b[j] = (a[j+1] - a[j])/h[j] - h[j]*(c[j+1] + 2*c[j])/3
    d[j] = (c[j+1] - c[j]) / (3*h[j])

# Resultados
print("c =", c)
print("b =", b)
print("d =", d)
```

```python
# Polinomios del Spline

# Se construye la expresión del spline cúbico en el intervalo correspondeinte con i=0,1,2
for i in range(n):
    expr = f"S_{i}(x) = {a[i]:.9f} + {b[i]:.9f}(x - {x[i]}) + {c[i]:.9f}(x - {x[i]})^2 + {d[i]:.9f}(x - {x[i]})^3"
    display(Math(expr)) # se imprime el resultado en formato latex 
```

El spline cúbico natural $S(x)$ está dado por:

$$
S(x)=
\begin{cases}

1.000000000 - 0.757359313(x-0.0) + 0.000000000(x-0.0)^2 - 6.627416998(x-0.0)^3, & 0 \le x \le 0.25 \\

0.707106781 - 2.000000000(x-0.25) - 4.970562748(x-0.25)^2 + 6.627416998(x-0.25)^3, & 0.25 < x \le 0.5 \\

0.000000000 - 3.242640687(x-0.5) + 0.000000000(x-0.5)^2 + 6.627416998(x-0.5)^3, & 0.5 < x \le 0.75 \\

-0.707106781 - 2.000000000(x-0.75) + 4.970562748(x-0.75)^2 - 6.627416998(x-0.75)^3, & 0.75 < x \le 1

\end{cases}
$$

```python
# Se usara S_1 para aproximar la 2.da derivada valuada en 0.5

# Variable simbólica
x = sp.symbols('x')

# Definir S1(x)
S1 = 0.707106781 - 2*(x - 0.25) - 4.970562748*(x - 0.25)**2 + 6.627416998*(x - 0.25)**3

# Segunda derivada
S1_2 = sp.diff(S1, x, 2)

# Evaluar en x = 0.5
valor = S1_2.subs(x, 0.5)

# Mostrar resultados
display(Math(r"S_1''(x) = " + sp.latex(sp.simplify(S1_2))))
display(Math(rf"S_1''(0.5) \approx {float(valor):.7f}"))
```
