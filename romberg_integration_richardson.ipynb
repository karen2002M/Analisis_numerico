```python
# Librerias 

import numpy as np
import pandas as pd
import sympy as sp
from IPython.display import display, Math

# Configuración numérica (9 dígitos)
np.set_printoptions(precision=9, suppress=True)
pd.set_option('display.precision', 9)
sp.init_printing(use_latex=True)
DIGITOS = 9
```

```python
# Función dada:
# f(x) = sqrt(1 + cos²(x))
def f(x):
    return np.sqrt(1 + np.cos(x)**2)

# Intervalo de integración
a = 0
b = 48
```

```python
# Implementación del método de Romberg

def romberg(f, a, b, n_max):
    """
    Construye la tabla de Romberg hasta R[n_max, n_max].

    Parámetros
    f :  Función a integrar.
    a, b : Límites de integración
    n_max : Nivel máximo de refinamiento.

    Retorna
    R : Tabla triangular de Romberg.
    """

    # Inicialización de la tabla
    R = np.zeros((n_max, n_max))

    # Primera entrada
    h = b - a  # Tamaño de paso
    R[0, 0] = (h / 2) * (f(a) + f(b)) # Primera aproximación trapezoidal (trapecio simple) correspondeinte a R_{11}

    # Construcción de la primera columna:
    for i in range(1, n_max):

        h /= 2

        # Nuevos nodos intermedios
        suma = 0

        for k in range(1, 2**i, 2):
            suma += f(a + k * h)

        # Fórmula recursiva del trapecio compuesto
        R[i, 0] = 0.5 * R[i - 1, 0] + h * suma

        # Fórmula de extrapolación de Richardson
        for j in range(1, i + 1):

            R[i, j] = (
                R[i, j - 1]
                + (R[i, j - 1] - R[i - 1, j - 1]) / (4**j - 1)
            )

    return R
```

```python
#Construcción y visualización de la tabla de Romberg

# Nivel máximo 
n_max = 10

# Construcción de la tabla
R = romberg(f, a, b, n_max)
indices = [f"R{i+1}" for i in range(n_max)]
columnas = [f"{j+1}" for j in range(n_max)]
tabla_romberg = pd.DataFrame(R, index=indices, columns=columnas)

# Redondeo 
tabla_romberg = tabla_romberg.round(DIGITOS)

# Mostrar tabla
display(tabla_romberg)
```

**e.**  
  
La dificultad en la integración de Romberg radica en que el intervalo de integración considerado es relativamente grande. Por lo cual la regla trapezoidal compuesta puede tomar áreas de más o de menos, debido a que la función dada oscila varias veces en dicho intervalo con período $\pi$.

Una forma de obtener mejores aproximaciones consiste en separar la integral en varios intervalos de integración asociados a cada período de la función. De esta manera, las aproximaciones trapezoidales se realizan sobre intervalos más pequeños. Lo cual mejora considerablemente la precisión y estabilidad numérica del método de Romberg.  
  
La separación de la integral puede realizarse como:  

$$ \int_{0}^{48}\sqrt{1+\cos^2(x)}\,dx=
\int_{0}^{\pi}\sqrt{1+\cos^2(x)}\,dx
+\int_{\pi}^{2\pi}\sqrt{1+\cos^2(x)}\,dx
+\cdots+
\int_{14\pi}^{15\pi}\sqrt{1+\cos^2(x)}\,dx
+\int_{15\pi}^{48}\sqrt{1+\cos^2(x)}\,dx$$  
  
Por periodicidad,  
$$
\int_{0}^{48}\sqrt{1+\cos^2(x)}\,dx=
15\int_{0}^{\pi}\sqrt{1+\cos^2(x)}\,dx
+
\int_{15\pi}^{48}\sqrt{1+\cos^2(x)}\,dx.
$$


```python
# Replanteamiento usando periodicidad 

# Integral en un periodo completo 
R_periodo = romberg(f, 0, np.pi, 10)

I_periodo = R_periodo[9,9]

# Integral restante en el intervalo [15\pi, 48]
R_restante = romberg(f, 15*np.pi, 48, 10)

I_restante = R_restante[9,9]

# Integral total
integral_e = 15 * I_periodo + I_restante

# Resultado
display(Math(fr"\int_0^{{48}} \sqrt{{1+\cos^2(x)}}\,dx \approx {integral_e:.9f}"))
```

```python
# Error de la aproximación usando separación por períodos

error_e = abs(valor_comp - integral_e)

display(Math(fr"E_e = "fr"\left|{valor_comp:.9f} - {integral_e:.9f}\right| "fr"= {error_e:.9f}"))

# Comparación con el error del inciso d el cual fue la mejor approx.
display(Math(fr"E_e < E_d \;\text{{ es }}\; {error_e < error_d}"))
```

Aunque la reformulación aprovecha la periodicidad de la función y reduce las oscilaciones locales en cada subintervalo, en este caso la aproximación obtenida directamente mediante $R_{10,10}$ presentó un menor error absoluto. Esto se debe a que es la aproximación más refinada es decir, utiliza un mayor número de particiones.
